```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Define model hyperparameters
VOCAB_SIZE = 10000  # Example vocabulary size for text data
EMBEDDING_DIM = 64
LSTM_UNITS = 64
NUM_CLASSES = 2     # Binary classification example

# Build the LSTM model
lstm_model = models.Sequential([
    # Input layer for sequences of integers (e.g., text token indices)
    layers.Input(shape=(None,), dtype='int32'),

    # Embedding layer to convert word indices to dense vectors
    layers.Embedding(input_dim=VOCAB_SIZE, output_dim=EMBEDDING_DIM),

    # LSTM layer to process sequential dependencies
    layers.LSTM(LSTM_UNITS, return_sequences=False),

    # Fully connected layers for classification
    layers.Dense(64, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(NUM_CLASSES, activation='softmax' if NUM_CLASSES > 1 else 'sigmoid')
])

# Compile the model
lstm_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy' if NUM_CLASSES > 1 else 'binary_crossentropy',
    metrics=['accuracy']
)

# Display model architecture
lstm_model.summary()
```

```python
import numpy as np

# Generate mock sequence data (e.g., 100 samples, each of length 20)
num_samples = 100
sequence_length = 20

X_train = np.random.randint(0, VOCAB_SIZE, size=(num_samples, sequence_length))
y_train = np.random.randint(0, NUM_CLASSES, size=(num_samples,))

# Train the LSTM model on mock data
print("Training the LSTM model...")
history = lstm_model.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=16,
    validation_split=0.2
)

# Print final training results
print("\nTraining completed successfully!")
```

```python
# Define model hyperparameters for the GRU model
GRU_UNITS = 64

# Build the GRU model
gru_model = models.Sequential([
    # Input layer for sequences of integers
    layers.Input(shape=(None,), dtype='int32'),

    # Embedding layer to convert word indices to dense vectors
    layers.Embedding(input_dim=VOCAB_SIZE, output_dim=EMBEDDING_DIM),

    # GRU layer to process sequential dependencies
    layers.GRU(GRU_UNITS, return_sequences=False),

    # Fully connected layers for classification
    layers.Dense(64, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(NUM_CLASSES, activation='softmax' if NUM_CLASSES > 1 else 'sigmoid')
])

# Compile the GRU model
gru_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy' if NUM_CLASSES > 1 else 'binary_crossentropy',
    metrics=['accuracy']
)

# Display model architecture
gru_model.summary()
```

```python
# Train the GRU model on the same mock data
print("Training the GRU model...")
gru_history = gru_model.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=16,
    validation_split=0.2
)
print("\nGRU Training completed successfully!")
```
