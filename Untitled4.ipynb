```python
!pip install --upgrade google-genai
import os
from google import genai
from google.colab import userdata
api_key=userdata.get('lufy')
os.environ['lufy'] = api_key
client = genai.Client(api_key=api_key)
prompt = "Explain quantum physics to a 5-year-old in exacly two sentences"
print("Contacting the LLM...\n")
response = client.models.generate_content(
    model="llama-3.3-70b-versatile",
    contents=prompt
)
print(response.text)
```
