```python
# ============================================================
# LANGGRAPH BASED LLM PROJECT
# State + Node + Route + Graph
# Google Colab - No API Key
# ============================================================

!pip install -q langgraph

from typing import TypedDict
from langgraph.graph import StateGraph, START, END


# ------------------------------------------------------------
# 1. STATE
# ------------------------------------------------------------

class GraphState(TypedDict):
    question: str
    category: str
    answer: str


# ------------------------------------------------------------
# 2. ROUTER NODE
# ------------------------------------------------------------

def router(state):
    question = state["question"].lower()

    if "artificial intelligence" in question or " ai" in question:
        category = "AI"

    elif "python" in question:
        category = "Python"

    elif "machine learning" in question:
        category = "Machine Learning"

    else:
        category = "General"

    return {"category": category}


# ------------------------------------------------------------
# 3. AI NODE
# ------------------------------------------------------------

def ai_node(state):
    return {
        "answer": (
            "Artificial Intelligence (AI) is a branch of "
            "computer science that enables machines to perform "
            "tasks that normally require human intelligence."
        )
    }


# ------------------------------------------------------------
# 4. PYTHON NODE
# ------------------------------------------------------------

def python_node(state):
    return {
        "answer": (
            "Python is a high-level programming language used "
            "for software development, data analysis, AI and ML."
        )
    }


# ------------------------------------------------------------
# 5. MACHINE LEARNING NODE
# ------------------------------------------------------------

def ml_node(state):
    return {
        "answer": (
            "Machine Learning is a branch of AI that allows "
            "computers to learn patterns from data and make "
            "predictions or decisions."
        )
    }


# ------------------------------------------------------------
# 6. GENERAL NODE
# ------------------------------------------------------------

def general_node(state):
    return {
        "answer": (
            "This question is processed by the general "
            "LLM node in the LangGraph."
        )
    }


# ------------------------------------------------------------
# 7. ROUTE FUNCTION
# ------------------------------------------------------------

def select_node(state):

    if state["category"] == "AI":
        return "ai"

    if state["category"] == "Python":
        return "python"

    if state["category"] == "Machine Learning":
        return "ml"

    return "general"


# ------------------------------------------------------------
# 8. CREATE GRAPH
# ------------------------------------------------------------

graph = StateGraph(GraphState)

graph.add_node("router", router)
graph.add_node("ai", ai_node)
graph.add_node("python", python_node)
graph.add_node("ml", ml_node)
graph.add_node("general", general_node)


# ------------------------------------------------------------
# 9. CONNECT NODES
# ------------------------------------------------------------

graph.add_edge(START, "router")

graph.add_conditional_edges(
    "router",
    select_node,
    {
        "ai": "ai",
        "python": "python",
        "ml": "ml",
        "general": "general"
    }
)

graph.add_edge("ai", END)
graph.add_edge("python", END)
graph.add_edge("ml", END)
graph.add_edge("general", END)


# ------------------------------------------------------------
# 10. COMPILE
# ------------------------------------------------------------

app = graph.compile()


# ------------------------------------------------------------
# 11. RUN AUTOMATICALLY
# ------------------------------------------------------------

result = app.invoke({
    "question": "What is Artificial Intelligence?",
    "category": "",
    "answer": ""
})


# ------------------------------------------------------------
# 12. DISPLAY OUTPUT
# ------------------------------------------------------------

print("=" * 50)
print("       LANGGRAPH LLM PROJECT")
print("=" * 50)

print("Question :", result["question"])
print("Route    :", result["category"])
print("Answer   :", result["answer"])

print("=" * 50)
print("          EXECUTION COMPLETED")
print("=" * 50)
```

```python
# ============================================================
# LANGGRAPH: TWO LLM NODES WITH ANSWER CHECKER
# Google Colab
# No API Key | No Input | No Files
# ============================================================

!pip -q install langgraph

from typing import TypedDict
from langgraph.graph import StateGraph, START, END


# ============================================================
# 1. STATE
# ============================================================

class State(TypedDict):
    question: str
    llm1_answer: str
    llm2_check: str
    final_answer: str


# ============================================================
# 2. FIRST LLM NODE
# ============================================================

def llm1_node(state):

    question = state["question"]

    # First LLM generates an answer
    if "artificial intelligence" in question.lower():
        answer = (
            "Artificial Intelligence is a branch of computer "
            "science that enables machines to perform tasks "
            "that normally require human intelligence."
        )

    else:
        answer = "This is the answer generated by LLM 1."

    return {
        "llm1_answer": answer
    }


# ============================================================
# 3. SECOND LLM - ANSWER CHECKER
# ============================================================

def llm2_checker(state):

    answer = state["llm1_answer"]

    # Second LLM checks the first answer
    if len(answer) > 20:
        check = "LLM 2: The answer is relevant and acceptable."
    else:
        check = "LLM 2: The answer needs improvement."

    return {
        "llm2_check": check
    }


# ============================================================
# 4. FINAL NODE
# ============================================================

def final_node(state):

    return {
        "final_answer":
            state["llm1_answer"]
            + "\n\n"
            + state["llm2_check"]
    }


# ============================================================
# 5. CREATE GRAPH
# ============================================================

workflow = StateGraph(State)

workflow.add_node("LLM_1", llm1_node)
workflow.add_node("LLM_2_Checker", llm2_checker)
workflow.add_node("Final", final_node)


# ============================================================
# 6. CREATE ROUTE
# ============================================================

workflow.add_edge(START, "LLM_1")
workflow.add_edge("LLM_1", "LLM_2_Checker")
workflow.add_edge("LLM_2_Checker", "Final")
workflow.add_edge("Final", END)


# ============================================================
# 7. COMPILE
# ============================================================

app = workflow.compile()


# ============================================================
# 8. RUN AUTOMATICALLY
# ============================================================

result = app.invoke({
    "question": "What is Artificial Intelligence?",
    "llm1_answer": "",
    "llm2_check": "",
    "final_answer": ""
})


# ============================================================
# 9. DISPLAY OUTPUT
# ============================================================

print("=" * 60)
print("       LANGGRAPH TWO-LLM ANSWER CHECKER")
print("=" * 60)

print("\nQUESTION:")
print(result["question"])

print("\nLLM 1 ANSWER:")
print(result["llm1_answer"])

print("\nLLM 2 CHECK:")
print(result["llm2_check"])

print("\nFINAL ANSWER:")
print(result["final_answer"])

print("\n" + "=" * 60)
print("             PROJECT COMPLETED")
print("=" * 60)
```

```python
# ============================================================
# GOOGLE COLAB
# LangGraph + 2 API Keys + 2 LLMs
#
# Flow:
#
#                 START
#                   |
#                   v
#                ROUTER
#               /      \
#              /        \
#             v          v
#         LLM 1        LLM 2
#             \          /
#              \        /
#               v      v
#                  END
# ============================================================


# ------------------------------------------------------------
# 1. INSTALL
# ------------------------------------------------------------

!pip -q install -U langgraph langchain-openai


# ------------------------------------------------------------
# 2. IMPORTS
# ------------------------------------------------------------

import os
from getpass import getpass
from typing import TypedDict, Literal

from langchain_openai import ChatOpenAI
from langgraph.graph import StateGraph, START, END


# ------------------------------------------------------------
# 3. TWO API KEYS
# ------------------------------------------------------------

print("Enter API KEY 1")
api_key_1 = getpass("API Key 1: ").strip()

print("\nEnter API KEY 2")
api_key_2 = getpass("API Key 2: ").strip()


# ------------------------------------------------------------
# 4. TWO LLMs
# ------------------------------------------------------------

llm_1 = ChatOpenAI(
    model="gpt-5",
    temperature=0,
    api_key=api_key_1
)

llm_2 = ChatOpenAI(
    model="gpt-5",
    temperature=0,
    api_key=api_key_2
)


# ------------------------------------------------------------
# 5. STATE
# ------------------------------------------------------------

class State(TypedDict):
    messages: list
    model: str


# ------------------------------------------------------------
# 6. ROUTER
# ------------------------------------------------------------

def route(
    state: State
) -> Literal["llm_1", "llm_2"]:

    # Get latest user message
    message = state["messages"][-1][1]

    message = message.lower()

    # Example routing logic
    #
    # Coding questions → LLM 1
    # General questions → LLM 2

    coding_words = [
        "code",
        "python",
        "javascript",
        "programming",
        "bug",
        "error",
        "function",
        "api"
    ]

    if any(word in message for word in coding_words):
        return "llm_1"

    return "llm_2"


# ------------------------------------------------------------
# 7. LLM 1 NODE
# ------------------------------------------------------------

def llm_1_node(state: State):

    response = llm_1.invoke(
        state["messages"]
    )

    return {
        "messages": state["messages"] + [response],
        "model": "LLM 1"
    }


# ------------------------------------------------------------
# 8. LLM 2 NODE
# ------------------------------------------------------------

def llm_2_node(state: State):

    response = llm_2.invoke(
        state["messages"]
    )

    return {
        "messages": state["messages"] + [response],
        "model": "LLM 2"
    }


# ------------------------------------------------------------
# 9. CREATE GRAPH
# ------------------------------------------------------------

graph = StateGraph(State)


# ------------------------------------------------------------
# 10. ADD NODES
# ------------------------------------------------------------

graph.add_node(
    "llm_1",
    llm_1_node
)

graph.add_node(
    "llm_2",
    llm_2_node
)


# ------------------------------------------------------------
# 11. START → ROUTER
# ------------------------------------------------------------

graph.add_conditional_edges(
    START,
    route,
    {
        "llm_1": "llm_1",
        "llm_2": "llm_2"
    }
)


# ------------------------------------------------------------
# 12. LLMs → END
# ------------------------------------------------------------

graph.add_edge(
    "llm_1",
    END
)

graph.add_edge(
    "llm_2",
    END
)


# ------------------------------------------------------------
# 13. COMPILE GRAPH
# ------------------------------------------------------------

app = graph.compile()


# ------------------------------------------------------------
# 14. USER INPUT
# ------------------------------------------------------------

user_input = input(
    "\nAsk something: "
)


# ------------------------------------------------------------
# 15. RUN GRAPH
# ------------------------------------------------------------

result = app.invoke({
    "messages": [
        ("user", user_input)
    ],
    "model": ""
})


# ------------------------------------------------------------
# 16. OUTPUT
# ------------------------------------------------------------

print("\n========================================")
print("MODEL USED:", result["model"])
print("========================================")

print("\nAI:")
print(result["messages"][-1].content)
```
